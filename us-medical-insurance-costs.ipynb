```python
import pandas as pd

# Load the dataset
df = pd.read_csv('insurance.csv')

# Verify dimensions, data types, and check for missing values
print(f"Dataset Shape: {df.shape[0]} rows, {df.shape[1]} columns\n")
print("--- Data Information ---")
df.info()

print("\n--- Missing Values ---")
print(df.isnull().sum())

print("\n--- First 5 Rows ---")
display(df.head())

print("\n--- Numerical Summary Statistics ---")
display(df.describe().round(2))
```

```python
# Save each dataset feature into its own dedicated variable
ages = df['age']
sexes = df['sex']
bmis = df['bmi']
num_children = df['children']
smoker_statuses = df['smoker']
regions = df['region']
charges = df['charges']

# Confirm variables are populated by checking lengths and sample values
print(f"Total entries per variable: {len(ages)}")
print(f"Sample - First Age: {ages[0]}, First Charge: ${charges[0]:.2f}")
```

```python
class InsuranceAnalysis:
    def __init__(self, ages, sexes, bmis, num_children, smoker_statuses, regions, charges):
        self.ages = ages
        self.sexes = sexes
        self.bmis = bmis
        self.num_children = num_children
        self.smoker_statuses = smoker_statuses
        self.regions = regions
        self.charges = charges

    def analyze_ages(self):
        """Calculates average, minimum, and maximum age."""
        avg_age = sum(self.ages) / len(self.ages)
        return {"mean_age": round(avg_age, 2), "min_age": min(self.ages), "max_age": max(self.ages)}

    def analyze_regions(self):
        """Calculates counts and percentages for each unique region."""
        total = len(self.regions)
        counts = {}
        for region in self.regions:
            counts[region] = counts.get(region, 0) + 1
        return {region: {"count": count, "percentage": round((count / total) * 100, 2)} 
                for region, count in counts.items()}

    def analyze_smoker_charges(self):
        """Compares average charges between smokers and non-smokers."""
        smoker_costs = [c for c, s in zip(self.charges, self.smoker_statuses) if s == 'yes']
        non_smoker_costs = [c for c, s in zip(self.charges, self.smoker_statuses) if s == 'no']

        avg_smoker = sum(smoker_costs) / len(smoker_costs)
        avg_non_smoker = sum(non_smoker_costs) / len(non_smoker_costs)

        return {
            "avg_smoker_cost": round(avg_smoker, 2),
            "avg_non_smoker_cost": round(avg_non_smoker, 2),
            "difference": round(avg_smoker - avg_non_smoker, 2)
        }

    def create_patient_dictionary(self):
        """Creates a consolidated dictionary for all patient records."""
        patient_data = {}
        for i in range(len(self.ages)):
            patient_data[i] = {
                "age": self.ages[i],
                "sex": self.sexes[i],
                "bmi": self.bmis[i],
                "children": self.num_children[i],
                "smoker": self.smoker_statuses[i],
                "region": self.regions[i],
                "charges": self.charges[i]
            }
        return patient_data

    def generate_summary_dictionary(self):
        """Compiles all high-level metrics into a nested dictionary."""
        return {
            "demographics": {
                "total_patients": len(self.ages),
                "average_age": self.analyze_ages()["mean_age"],
                "regional_breakdown": self.analyze_regions()
            },
            "financials": {
                "smoker_impact": self.analyze_smoker_charges(),
                "average_overall_charge": round(sum(self.charges) / len(self.charges), 2)
            }
        }

    def analyze_feature_influence(self):
        """Calculates Pearson correlation coefficient for numerical/binary features against charges."""
        smoker_num = [1 if s == 'yes' else 0 for s in self.smoker_statuses]
        sex_num = [1 if s == 'male' else 0 for s in self.sexes]
        
        def pearson_r(x, y):
            n = len(x)
            mean_x, mean_y = sum(x)/n, sum(y)/n
            cov = sum((xi - mean_x) * (yi - mean_y) for xi, yi in zip(x, y))
            std_x = (sum((xi - mean_x)**2 for xi in x)) ** 0.5
            std_y = (sum((yi - mean_y)**2 for yi in y)) ** 0.5
            return round(cov / (std_x * std_y), 4)

        return {
            "smoker_correlation": pearson_r(smoker_num, self.charges),
            "age_correlation": pearson_r(self.ages, self.charges),
            "bmi_correlation": pearson_r(self.bmis, self.charges),
            "children_correlation": pearson_r(self.num_children, self.charges),
            "sex_correlation": pearson_r(sex_num, self.charges)
        }

# Execute class methods
analysis = InsuranceAnalysis(ages, sexes, bmis, num_children, smoker_statuses, regions, charges)

print("Age Statistics:", analysis.analyze_ages())
print("\nRegional Distribution:", analysis.analyze_regions())
print("\nSmoker vs Non-Smoker Costs:", analysis.analyze_smoker_charges())
```

## Extension 3: Exploring Dataset Bias & Limitations

When evaluating this dataset for real-world application or predictive risk modeling, several structural biases and sampling limitations must be considered:

### 1. Severe Subgroup Underrepresentation (Sampling Bias)
* **Underweight Smoker Representation:** Out of 1,338 records, only **21 individuals** are classified as "Underweight" ($\text{BMI} < 18.5$), and only **5** of those are smokers. 
* **Impact:** Statistical estimates or predictive models trained on this data will have low statistical power and high variance when predicting outcomes for underweight individuals.

### 2. Omission of Critical Health & Clinical Metrics (Confounding Variables)
* **Missing Clinical Markers:** The dataset contains no medical history, blood pressure readings, cholesterol levels, family history of chronic illnesses, pre-existing conditions, or lifestyle factors (e.g., alcohol consumption, exercise routines).
* **Impact:** Attributing higher insurance costs solely to age, BMI, and smoking status risks oversimplifying risk models. Two individuals with identical age and BMI could have radically different medical expenses depending on underlying conditions not captured in this file.

### 3. Coarse Geographic & Socioeconomic Granularity
* **Broad Regional Buckets:** Location is recorded across only four broad quadrants (`northeast`, `northwest`, `southeast`, `southwest`).
* **Impact:** Healthcare pricing varies drastically by state regulations, local cost of living, urban vs. rural healthcare access, and regional hospital network monopolies. Lumping entire regions into single categories ignores micro-economic variations in medical pricing.

### 4. Selection & Survivorship Bias
* **Active Policyholders Only:** This dataset only captures individuals who currently hold medical insurance policies.
* **Impact:** Uninsured individuals—who may face higher health risks or abstain from insurance due to cost barriers—are completely excluded. Models trained on this dataset cannot be generalized to the broader national population.
